```python
# Imports
import os
from qiskit import QuantumCircuit, qpy
```

```python
def maslov_cost(file_path):
  """
  Calculates the maslov cost of a circuit

  Args:
    file_path (str): Path to the circuit file (either .qpy or .qasm)

  Returns:
    cost (int): Maslov cost of the circuit
  """
  if os.path.splitext(file_path)[1] == ".qpy":
    with open(file_path, 'rb') as fd:
      circuits = qpy.load(fd)
    qc = circuits[0]
  elif os.path.splitext(file_path)[1] == ".qasm":
    qc = QuantumCircuit.from_qasm_file(file_path)
  else:
    print("File type not supported")
    return

  cost = 0

  for instruction in qc.data:
    op = instruction.operation
    name = op.name

    if name == 'x':
      cost += 1
    elif name in ['cx', 'ccx', 'mcx']:
      cost += 2**(op.num_qubits)-3
    elif name == 'swap':
          cost += 3
    elif name == 'swat':
      cost += 8  # 3 CNOT + 3-bit Toffoli gate
    elif name == 'swat1':  # Reduced SWAT gate
      cost += 2

  return cost
```

# Run on Benchmarks

```python
# Circuit type ("lattice", "esop", or "bdd")
circuit_type = "lattice"

# File type ("qasm" or "qpy")
file_type = "qasm"
```

```python
# Calculate for a single circuit
benchmark = "sam_ex1" # <--- Change

print(f"{benchmark}: {maslov_cost(f"./circuits/{circuit_type}_circuits/{benchmark}-{circuit_type}.{file_type}")}")
```

```python
# Calculate for a all circuits
benchmarks = ["sam_ex1", "sam_ex2", "sam_ex3", "sam_ex4", "sam_ex5",
              "rd53f1", "rd53f2", "rd73f1", "rd73f3", "rd84f1", "rd84f3", "rd84f4",
              "sym6_63", "sym9_71", "sym10_207", "co14_135"]

for benchmark in benchmarks:
    print(f"{benchmark}: {maslov_cost(f"./circuits/{circuit_type}_circuits/{benchmark}-{circuit_type}.{file_type}")}")
```
